```python
from typing import TypedDict, Literal
from langgraph.graph import StateGraph, START, END
from IPython.display import Image, display

class CurrencyState(TypedDict):
    amount_inr: float
    target_currency: Literal['USD', 'EUR']
    converted_usd: float
    converted_eur: float


def convert_to_usd(state: CurrencyState) -> CurrencyState:
    state['converted_usd'] = state['amount_inr'] / 85.0
    return state

def convert_to_eur(state: CurrencyState) -> CurrencyState:
    state['converted_eur'] = state['amount_inr'] / 91.0
    return state

def choose_conversion(state: CurrencyState) -> str:
    return state['target_currency']
```

```python
# build the graph by connecting nodes by edges

builder = StateGraph(CurrencyState)

builder.add_node("convert_to_usd", convert_to_usd)
builder.add_node("convert_to_eur", convert_to_eur)

builder.add_conditional_edges(
    START, choose_conversion,
    {
        "USD": "convert_to_usd",
        "EUR": "convert_to_eur"
    })
builder.add_edge(["convert_to_usd", "convert_to_eur"], END)

graph = builder.compile()
```

```python
# display the graph
display(Image(graph.get_graph().draw_mermaid_png()))
```

```python
# invoke the graph with initial state
graph.invoke({"amount_inr": 1000, "target_currency": "EUR"})
```
