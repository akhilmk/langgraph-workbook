```python
from dotenv import load_dotenv
from typing import TypedDict, Annotated
from langgraph.graph import StateGraph, START, END, add_messages
from langchain.chat_models import init_chat_model
from IPython.display import Image, display

load_dotenv()
```

```python
# initialize the LLM
llm = init_chat_model("llama-3.1-8b-instant", model_provider="groq")
```

```python
# graph state
class ChatState(TypedDict):
    messages: Annotated[list, add_messages]

# graph node function
def chatbot(state: ChatState)-> ChatState:
    return {"messages": [llm.invoke(state["messages"])]}

# build the graph by connecting nodes by edges
builder = StateGraph(ChatState)
builder.add_node("chatbot_node", chatbot)

builder.add_edge(START, "chatbot_node")
builder.add_edge("chatbot_node", END)

graph=builder.compile()

display(Image(graph.get_graph().draw_mermaid_png()))
```

```python
# calling LLM inline
msg = {"role":"user", "content": "what is the capital of India?"}
respState = graph.invoke({"messages": [msg]})
respState["messages"]
```

```python
# calling LLM for recent data. Need to use tool calls for this to work.
msg = {"role":"user", "content": "what is the price of GOOGL stock?"}
respState = graph.invoke({"messages": [msg]})
respState["messages"]
```

```python
# chatbot listening to user input in a loop
chatState = None
while True:
    user_input = input("User: ")
    if user_input.lower() in ["exit", "quit"]:
        break
    chatState = graph.invoke({"messages": [{"role": "user", "content": user_input}]})
    print("LLM:", chatState["messages"][-1].content)
```
